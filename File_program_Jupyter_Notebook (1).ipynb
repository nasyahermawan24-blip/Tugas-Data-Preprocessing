```python

import pandas as pd

# Membaca dataset
df = pd.read_csv('/content/drive/MyDrive/DataSharing/dataset_penggunaan_media_sosial_mahasiswa.csv')


# Melihat jumlah missing value
print("Jumlah Missing Value:")
print(df.isnull().sum())

# Melihat total missing value
print("\nTotal Missing Value:")
print(df.isnull().sum().sum())
```

**Memperbaiki nilai yang salah format**

```python

# Memperbaiki nilai teks pada Academic_Score
df["Academic_Score"] = df["Academic_Score"].replace(
    "delapan puluh", "80"
)

# Mengubah kolom menjadi numerik
df["Academic_Score"] = pd.to_numeric(
    df["Academic_Score"], errors="coerce"
)

# Memeriksa hasil
print(df["Academic_Score"].dtype)
print(df["Academic_Score"].describe())
```

**Menghitung statistik deskriptif**

```python

# Menghitung median setiap variabel numerik
print("Median Daily Social Media Hours:",
      df["Daily_Social_Media_Hours"].median())

print("Median Academic Score:",
      df["Academic_Score"].median())

# Menghitung modus variabel kategorik
print("Modus Gender:",
      df["Gender"].mode()[0])

print("Modus Most Used Platform:",
      df["Most_Used_Platform"].mode()[0])
```

**Mengisi Missing Value**

```python

# Mengisi missing value kategorik menggunakan modus
df["Gender"] = df["Gender"].fillna(
    df["Gender"].mode()[0]
)

df["Most_Used_Platform"] = df["Most_Used_Platform"].fillna(
    df["Most_Used_Platform"].mode()[0]
)

# Mengisi missing value numerik menggunakan median
df["Daily_Social_Media_Hours"] = df["Daily_Social_Media_Hours"].fillna(
    df["Daily_Social_Media_Hours"].median()
)

df["Academic_Score"] = df["Academic_Score"].fillna(
    df["Academic_Score"].median()
)

# Memeriksa kembali missing value
print(df.isnull().sum())
```

**Dataset yang sudah melalui proses Missing Value Handling dan Duplicate Data Handling.**

```python

# 1. Membersihkan spasi berlebih
df["Student_ID"] = df["Student_ID"].str.strip()
df["Gender"] = df["Gender"].str.strip()
df["Most_Used_Platform"] = df["Most_Used_Platform"].str.strip()
df["Study_Program"] = df["Study_Program"].str.strip()
df["Primary_Purpose"] = df["Primary_Purpose"].str.strip()

# 2. Menyamakan penulisan Gender
df["Gender"] = (
    df["Gender"].str.lower()
    .replace({
        "perempuan": "Perempuan",
        "laki-laki": "Laki-laki"
    })
)

# 3. Menyamakan penulisan platform
df["Most_Used_Platform"] = (
    df["Most_Used_Platform"].str.lower()
    .replace({
        "tiktok": "TikTok",
        "tik tok": "TikTok",
        "whatsapp": "WhatsApp",
        "instagram": "Instagram",
        "facebook": "Facebook",
        "youtube": "YouTube",
        "x": "X"
    })
)

# 4. Memperbaiki nama program studi
df["Study_Program"] = df["Study_Program"].replace({
    "Sistem Informsi": "Sistem Informasi"
})

# 5. Memeriksa hasil
print("Kategori Gender:")
print(df["Gender"].unique())

print("\nKategori Platform:")
print(df["Most_Used_Platform"].unique())

print("\nKategori Program Studi:")
print(df["Study_Program"].unique())

print("\nKategori Tujuan Penggunaan:")
print(df["Primary_Purpose"].unique())
```

**Inconsistent Data Handling**
Tahap ini dilakukan untuk memperbaiki data yang memiliki perbedaan penulisan, penggunaan huruf besar dan kecil, spasi berlebih, serta kesalahan pengetikan. Tujuannya agar setiap kategori memiliki format yang konsisten dan tidak dianggap sebagai kategori yang berbeda.

```python

# Menampilkan kategori sebelum dibersihkan
print("Gender sebelum dibersihkan:")
print(df["Gender"].unique())

print("\nPlatform sebelum dibersihkan:")
print(df["Most_Used_Platform"].unique())

print("\nProgram Studi sebelum dibersihkan:")
print(df["Study_Program"].unique())

print("\nTujuan Penggunaan sebelum dibersihkan:")
print(df["Primary_Purpose"].unique())
```

```python

# Menghapus spasi berlebih pada kolom kategorik
kolom_kategorik = [
    "Gender",
    "Study_Program",
    "Most_Used_Platform",
    "Primary_Purpose"
]

for kolom in kolom_kategorik:
    df[kolom] = df[kolom].str.strip()

# Menyamakan penulisan Gender
df["Gender"] = df["Gender"].replace({
    "perempuan": "Perempuan",
    "laki-laki": "Laki-laki"
})

# Menyamakan penulisan platform
df["Most_Used_Platform"] = df["Most_Used_Platform"].replace({
    "tiktok": "TikTok",
    "Tik Tok": "TikTok"
})

# Memperbaiki kesalahan pengetikan program studi
df["Study_Program"] = df["Study_Program"].replace({
    "Sistem Informsi": "Sistem Informasi"
})

# Menampilkan hasil setelah dibersihkan
print("Gender:", df["Gender"].unique())
print("Platform:", df["Most_Used_Platform"].unique())
print("Program Studi:", df["Study_Program"].unique())
print("Tujuan Penggunaan:", df["Primary_Purpose"].unique())
```

**Data Type Conversion**

```python

# Memperbaiki nilai teks pada kolom Age
df["Age"] = df["Age"].replace({
    "dua puluh": "20"
})

# Mengubah kolom menjadi tipe numerik
kolom_numerik = [
    "Age",
    "Daily_Social_Media_Hours",
    "Sleep_Hours_Per_Day",
    "Academic_Score",
    "Addiction_Score"
]

for kolom in kolom_numerik:
    df[kolom] = pd.to_numeric(df[kolom], errors="coerce")

# Memeriksa tipe data setelah konversi
df.info()
```

**Invalid Data Handling**

```python

# Memeriksa nilai yang tidak valid
print("Age di atas 100:")
display(df[df["Age"] > 100])

print("Durasi media sosial di atas 24 jam:")
display(df[df["Daily_Social_Media_Hours"] > 24])

print("Durasi tidur negatif:")
display(df[df["Sleep_Hours_Per_Day"] < 0])

print("Addiction Score di luar skala 1-10:")
display(df[
    (df["Addiction_Score"] < 1) |
    (df["Addiction_Score"] > 10)
])
```

```python
import numpy as np

# Mengubah nilai tidak valid menjadi missing value
df.loc[df["Age"] > 100, "Age"] = np.nan

df.loc[
    df["Daily_Social_Media_Hours"] > 24,
    "Daily_Social_Media_Hours"
] = np.nan

df.loc[
    df["Sleep_Hours_Per_Day"] < 0,
    "Sleep_Hours_Per_Day"
] = np.nan

df.loc[
    (df["Addiction_Score"] < 1) |
    (df["Addiction_Score"] > 10),
    "Addiction_Score"
] = np.nan

# Mengisi kembali missing value numerik dengan median
for kolom in kolom_numerik:
    df[kolom] = df[kolom].fillna(df[kolom].median())

print("Total missing value setelah perbaikan:",
      df.isnull().sum().sum())
```

**Outlier Handling**

```python

# Mendeteksi outlier menggunakan metode IQR
kolom_outlier = [
    "Age",
    "Daily_Social_Media_Hours",
    "Sleep_Hours_Per_Day",
    "Academic_Score",
    "Addiction_Score"
]

for kolom in kolom_outlier:
    Q1 = df[kolom].quantile(0.25)
    Q3 = df[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    jumlah_outlier = (
        (df[kolom] < batas_bawah) |
        (df[kolom] > batas_atas)
    ).sum()

    print(f"\nKolom: {kolom}")
    print("Batas bawah:", batas_bawah)
    print("Batas atas:", batas_atas)
    print("Jumlah outlier:", jumlah_outlier)
```
